```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('//kaggle/StressLevelDataset.csv')
```

```python

print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)
```

```python
df.info()
```

```python
print("Statistical Summary:")
display(df.describe())

print("\nMissing Values Check:")
print(df.isnull().sum())
```

```python
sns.histplot(df['anxiety_level'], color='teal', bins=20)

plt.title('Univariate Analysis: Distribution of Students Anxiety Levels', fontsize=10, )
plt.xlabel('Anxiety Level Score', fontsize=12)
plt.ylabel('Count', fontsize=10)



plt.show()
```

```python
sns.boxplot(x='stress_level', y='sleep_quality', data=df)


plt.title('Bivariate Analysis: Sleep Quality vs Stress Level', fontsize=15, fontweight='bold')
plt.xlabel('Stress Level (0=Low, 1=Medium, 2=High)', fontsize=12)
plt.ylabel('Sleep Quality Score', fontsize=12)

plt.show()
```

```python

selected_cols = ['stress_level', 'anxiety_level', 'depression', 'sleep_quality', 'self_esteem', 'academic_performance']


corr_matrix_small = df[selected_cols].corr()


plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix_small, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

plt.title('Multivariate Analysis', fontsize=14, fontweight='bold')
plt.show()
```
